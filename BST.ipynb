```python
class TreeNode:
  def __init__(self,node):
    self.node=node
    self.left=None
    self.right=None
```

```python
def BST(data):
  if(type(data)==tuple and len(data)==3):  
      node=TreeNode(data[1])
      node.left=BST(data[0])
      node.right=BST(data[2])
      return node

  elif(data == None):
    return None
  else:
    return TreeNode(data)
    
```

```python
BST(((4,3,1),5 ,(6,7,8)))
```
