Find First And Last Position of Element In Sorted Array(ascending)

```python
def binary_search(l,h,condition):
    while(l<=h):
        mid=(l+h)//2
        result=condition(mid)
        if(result=='found'):
            return mid
        elif(result == 'right'):
            l=mid+1
        elif(result =='left'):
            h=mid-1
    return -1


```

```python
def first_position(target,nums):
    def condition(mid):
        if(nums[mid]==target):
            if(mid>0 and nums[mid-1]==target):
                return 'left'
            else:
                return 'found'
        elif(nums[mid]>target):
            return 'left'
        else:
            return 'right'
    return binary_search(0,len(nums)-1,condition)
```

```python
def last_position(target,nums):
    def condition(mid):
        if(nums[mid]==target):
            if(mid<len(nums)-1 and nums[mid+1]==target):
                return 'right'
            else:
                return 'found'
        elif(nums[mid]>target):
            return 'left'
        else:
            return 'right'
    return binary_search(0,len(nums)-1,condition)
```

```python
def first_and_last_position(target,nums):
    return [first_position(target,nums), last_position(target,nums)]
```

```python

nums=[1,20,4,39,29,22,1,20,20,10,10,3]
nums.sort()
target=20
print(nums)
print(first_and_last_position(target,nums))
     
```
